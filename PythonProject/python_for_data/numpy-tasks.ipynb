## Weather Data

```python
temperatures = np.array([18, 21, 25, 17, 16, 23, 28, 30, 22, 19, 15, 24, 27, 20])
```

### Days above 25 degrees

```python
np.sum(temperatures > 25)
```

### Days below 20 degrees

```python
np.sum(temperatures < 20)
```

### Avg temp

```python
temperatures.mean()
```

### Hottest day

```python
temperatures.max()
```

### Coldest day

```python
temperatures.min()
```

### How many days exceeded the avg temp?

```python
np.sum(temperatures > temperatures.mean())
```

## Retails Sales

- Rows represent stores
- Columns represent weekly sales

```python
sales = np.array([[120, 150, 180, 200], [90, 110, 140, 160], [200, 220, 250, 300]])
```

### Total sales per store

```python
sales.sum(1)
```

### Total sales per week

```python
sales.sum(0)
```

### Overall sales

```python
sales.sum()
```

### Best performing store

```python
np.argmax(sales.sum(1))
```

### Best performing week (all stores)

```python
np.argmax(sales.sum(0))
```

### Avg sales per store

```python
print(np.mean(sales[0]))
print(np.mean(sales[1]))
print(np.mean(sales[2]))
```

## .reshape

Modify number of rows and columns in an array without affecting data (total number of elements must stay the same)

```python
temperatures.reshape(2, 7)
```

*1d array converted to 2d array, splitting temperatures by week*

## .transpose

Swap rows and columns of an array (i.e. turn 90 degrees)

```python
sales.transpose()
```

*Instead of rows representing stores and columns representing weekly sales as before, rows now represent weekly sales and columns represent stores*
